Loop  Control Statements

```python
## break
## break exits loop prematurely
for i in range(10):
    if(i==5):
        break
    print(i)
```

```python
## continue 
##skips the current operation and continues with the next
for i in range(10):
    if (i%2==0):
        continue
    print(i)
```

```python
## paas
## Pass statement is a null operation ,it does nothing

for i in range(5):
    if(i==3):
        print("the number is",i)
        pass
    print(i)
```

```python
## Nested loop- loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
def compute_GCD(x, y):  
   while(y): 
       x, y = y, x % y 
   return x 
  
a = 60
b = 48

print(compute_GCD(a,b))
```

```python
def compute_hcf(x, y):
# choose the smaller number
    if x > y:
        smaller = y
    else:
        smaller = x
    for i in range(1, smaller+1):
        if((x % i == 0) and (y % i == 0)):
            hcf = i 
    return hcf

num1 = 60
num2 = 48
print("The H.C.F. is", compute_hcf(num1, num2))
```

```python
def fibonacci(n):
    a = 0
    b = 1
    if n < 0:
        print("Incorrect input")
    elif n == 0:
        return a
    elif n == 1:
        return b
    else:
        for i in range(2,n+1):
            c = a + b
            a = b
            b = c
        return b
print(fibonacci(15))
```

```python
def splitArr(arr, n, k):
	for i in range(0, k):
		x = arr[0]
		for j in range(0, n-1):
			arr[j] = arr[j + 1]
		
		arr[n-1] = x
		
# main
arr = [12, 10, 5, 6, 52, 36]
n = len(arr)
position = 2

splitArr(arr, n, position)
for i in range(0, n):
	print(arr[i],end=" ")
```
