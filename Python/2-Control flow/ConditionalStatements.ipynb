Conditional Statements

```python
## If statement
age=20
if age>=18:
    print("You are allowed to vote in the election")
```

```python
age>=18
```

```python
## Else statement
## It executes when if condition in the if statement is false
age=16
if age>=18:
    print("you are allowed to vote")
else:
    print("you are a minor")
```

```python
## elif(else-if) statement : allows you to check multiple conditions
age=20
if age<13:
    print("YOu are a child")
elif age<18:
    print("you are a teenager")
else:
    print("you are an adult")
```

```python
## Nested conditional statement
## You can one or more if , elif, else conditions inside another if,elif ,else statement

## number even, odd, negative

num= int(input("Enter a number"))
if(num>=0):
    print("The number is positive")
    if(num %2 ==0):
        print("number is even")
    else:
        print("number is odd")
else:
    print("number is negative")
```

```python
## Practical examples
 
## determine if a year is a leap year using nested
year=int(input("Enter year"))
if(year%4==0):
    if(year %100==0):
        if(year%400==0):
            print(year," is not a leap year")
        else:
            print(year," is not a leap year")
    else:
        print(year," is a leap year")
else:
    print(year," is not a leap year")
```

```python
##Calculator
num1 = float(input("Enter first number"))
num2 = float(input("Enter second number"))
operator=input("Enter operator: +,-,*,/ :")

if(operator=="+"):
    result=num1+num2
elif(operator =="-"):
    result=num1-num2
elif(operator=="*"):
    result=num1*num2
elif(operator=="/"):
    result = num1/num2
else:
    print("Enter correct operator")
print(result)
```
