Sets

Built in data type used to store collection of unique items.

Unordered meaning they do not follow a specific order.

Do not allow duplicate elements.

```python
## Creating a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_set=set([1,2,3,4,5,6])
print(my_set)
```

```python
my_set=set([1,3,2,3,4,5,6])
print(my_set)
```

```python
## Basic set operation

## Adding and removing elements
my_set.add(8)
print(my_set)
my_set.add(8)
print(my_set)
```

```python
## Removing elements
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(9)
```

```python
## Check and remove element if present 
## Does not give error if not present
my_set.discard(10)
print(my_set)
```

```python
## Pop method- remove and return first element
removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
## Clear all the elements
my_set.clear()
print(my_set)
```

```python
## Set Membership test
my_set ={1,2,3,4,5}
print(2 in my_set)
print(10 in my_set)
```

```python
## Mathematical operations
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Union
union_set= set1.union(set2)
print(union_set)


##Intersection
intersection_set=set1.intersection(set2)
print(intersection_set)


##finds intersection elements and update it in set
set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Difference
print(set1.difference(set2))
## difference_update operation is also there similar to intersecion_update
```

```python
set2.difference(set1)
```

```python
## Symmatric difference- combines all unique elements in both sets
set1.symmetric_difference(set2)
```

```python
## Sets methos 
set1={1,2,3,4,5}
set2={3,4,5}

## is subset
print(set1.issubset(set2))

print(set1.issuperset(set2))
```

```python
## Removing duplicate elements from a list-- Just convert it to set
lst=[1,2,2,3,4,4,5]
set(lst)
```

```python
## Counting unique words in text
text="In this tutorial we are discussing about sets are"
word=text.split()
## Convert list of words to set to get unique words
unique_words=set(word)
print(unique_words)
print(len(unique_words))
```
