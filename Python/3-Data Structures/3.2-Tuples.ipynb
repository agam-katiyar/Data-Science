Tuples

Ordered  collections of items that are immutable.  They are similar to lists,but different in immutability.

```python
##Creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
## list converted to tuple
numbers=tuple([1,2,3,4,5,6])
numbers
```

```python
##Tuple comverted to list
list((1,2,3,4,5,6))
```

```python
mixed_tuple=(1,"agam",3.2,True)
print(mixed_tuple)
```

```python
##Accessing tuple elements
numbers
```

```python
print(numbers[0])
print(numbers[2])
print(numbers[-1])
```

```python
##Slicing
numbers[0:4]
```

```python
numbers[::-1]
```

##Tuple Operations

```python
##Concatenation
concatenation_tuple= numbers+mixed_tuple
print(concatenation_tuple)
```

```python
mixed_tuple*3
```

```python
numbers*3
```

```python
##Immmutable nature of tuples
lst=[1,2,3,4,5]
print(lst)
lst[1]=3 #lists are mutable
print(lst)

tuples=(1,2,3,4,5)
print(tuples)
tuples[0]=0 ##immutable
print(tuples)
```

```python
##Tuple methods
print(numbers.count(1)) ## occurence of numbers

print(numbers.index(3)) ## returns the index of first occurence of value 
```

## Packing and unpacking tuples


```python
## Packing
packed_tuple=1,"hello",3.4
print(packed_tuple)
```

```python
## Unpacking a tuple
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers=(1,2,3,4,5,6)
first, *second, third =numbers ## * select all elements till last except last element
print(first)
print(second)
print(third)
```

```python
## Nested tuple

## Nested list
list=[[1,2,3,4],[6,7,8,9],[1,"hello",54.6,"f"]]
print(list[0])
print(list[0][3])
print(list[0][0:3])
```

```python
nested_tuple=((1,2,3),('a','b','c'),(True,False))
## access the elements
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
