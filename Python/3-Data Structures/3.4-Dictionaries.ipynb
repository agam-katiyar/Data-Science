Dictionaries

Unnordered Collection of items.
Store data in key-value pairs.
Key must be unique and immutable(Strings,numbers,tuples).

```python
## Creating dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"agam", "age": "21","grade":34}
print(student)
print(type(student))
```

```python
## Unique key name
student={"name":"agam", "age": "21","name":34} ##name updates value to 34
print(student)
```

```python
## Accessing dictionary elements
student={"name":"agam", "age": "21","grade":'A'}
print(student)
```

```python
print(student['grade'])
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name')) return none if not available
print(student.get('last_name',"Not available")) ## providing default value
```

```python
## Modifying dictionary elements 
## Dictionaries are mutable,  can add,update or delete elements

print(student)
```

```python
student["age"]=33 ## Updating value
print(student)
student["address"]="India" ## adding new key & value
print(student)
```

```python
## Delete the key and value pair
del student['grade']
print(student)
```

```python
##Dictionary methods

keys=student.keys()  ## get all keys
print(keys)
values=student.values() ## get all values
print(values)


items=student.items()  ## get all key value pairs
print(items)
```

```python
## Shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="krish"
print(student)
print(student_copy)
```

```python
## Deep copy
student_copy1=student.copy()
print(student_copy1)
print(student)
```

```python
student["name"]="krish3"
print(student)
print(student_copy1)
```

```python
##Iterating over dictionaries
## Can use loops to iterate over dictionaries , keys,values,or items

##Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterating over values
for values in student.values():
    print(values)
```

```python
## Iterate over key- value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested dictionaries
students={
    "student1":{"name":"krish","age":23},
    "student2":{"name":"agam","age":24},
    "student3":{"name":"prince","age":26}
}
print(students)
```

```python
## Access nested dictionaries elements
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary comprehension
square={x:x**2 for x in range(5)}
print(square)
```

```python
## Conditional dictionary conprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical examples

## Use a dictionary to count frequency of elements in list
numbers=[1,2,2,2,3,3,4,4,4]
frequency={}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge 2 dictionaries

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merge_dict={**dict1,**dict2}
print(merge_dict)
```
