Inheritance

```python
#base class
class Human:
    def __init__(self,name, age):
        self.name = name
        self.age = age
    def gen(self):
        print(f"gender is human")

    def speak(self):
        print("Hello")

#derived class 
class Woman(Human):
    def __init__(self, name, age):
        super().__init__(name, age) ## Inheriting parent class attribute

    def gen(self):
        print("female")

    def speak(self):
        print("no")
        return super().speak()

    
```

```python
p2 = Woman("nik",45)
p2.gen()
```

```python
p2.speak()
```

1. Can an abstract method have a body/implementation in Python?

Answer: Yes. 
This surprises many candidates who come from Java or C#. In Python, you can write code inside an abstract method.Why do it? To provide a default or common behavior that child classes can invoke using super().

2. What happens if a child class forgets to implement an abstract method?

Answer: The script will still compile, but Python will raise a TypeError at runtime the exact moment you try to instantiate the child class. You cannot create an object of that child class until all abstract methods are implemented.

3. How does Python enforce abstract classes under the hood?

Answer: Python enforces abstract classes via a Metaclass called ABCMeta. When you inherit from ABC, you are implicitly setting metaclass=ABCMeta.During object instantiation, Python checks the __abstractmethods__ attribute of the class. If that set is not empty, Python prevents object creation.

4. Can you instantiate an abstract class if it has NO abstract methods?

Answer: Yes. If you create a class that inherits from ABC but you do not mark any method with @abstractmethod, Python treats it as a regular concrete class. You will be able to instantiate it freely.

5. How do Abstract Classes differ from Duck Typing in Python?

Answer:Duck Typing is Python's default philosophy: "If it walks like a duck and quacks like a duck, it's a duck." It does not check types; it just tries to call methods at runtime.Abstract Classes (ABCs) introduce Nominal Typing. They explicitly enforce structure before a method is ever called, making your code safer and self-documenting.

Encapsulation:

concept of wrapping data (cariables) and methods together as a single unit. It restrict direct access to some of object's components, which is a means of preventing accidental interference and misuse of data

```python
# we can perform encapsulation with getter and setter method 


# using public,private access modifier(( use single _ for protected and double _ before name for private variable))
class Person:
    def __init__(self,name,age,gender):
        self.__name = name ## private variable
        self.__age = age ## private
        self.gender = gender #public
```

```python
## Encapsulation with getter and setter
class Person:
    def __init__(self,name,age):
        self.__name = name ##private
        self.__age = age ## private

    ## getter method for accessing name
    def get_name(self):
        return self.__name
    
    ## setter method for name
    def set_name(self,name):
        self.__name = name

    def get_age(self):
        return self.__age
    
    def set_age(self,age):
        if(age>0):
            self.__age = age
        else:
            print("Age can't be negative")


person = Person("agam",20)

## accesss and modify private variablees using gettter and setter
print(person.get_name())
print(person.get_age())

person.set_age(21)
print(person.get_age())

person.set_age(-4)
```

Abstraction:
concept of hiding the complex implementation details and showing only necesssary features of an obejct.


```python
from abc import ABC, abstractmethod

## abstract claas
class Vehicle(ABC):
    def drive(self):
        print("Vehicle is used for driving")


    ## abstract method is implemented at other derived class 
    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):
    def start_engine(self):
        print("car engine started")

def operate_vehicle(vehicle):
    vehicle.start_engine()

car=Car()
operate_vehicle(car)
```

Operator Overloading

allows you to define behaviour for custom objects

```python
##Mathematical operation for vectors
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    
    def __add__(self,other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self,other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self,other):
        return Vector(self.x * other, self.y * other)

    def __eq__(self,other):
        return self.x == other.x and self.y == other.y
    
    def __repr__(self):
        return f"Vector({self.x},{self.y})"

## creating objects of vector class
v1=Vector(2,3)
v2=Vector(4,5)
print(v1+v2)
print(v1-v2)
print(v1*4)
```

```python
# Creating own iterator using __init__() -> (returns an iterator) and __next__() ->(returns a next value)
class CountUp:
    def __init__(self, limit):
        self.current = 1
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value
        else:
            raise StopIteration
```

```python
counter = CountUp(3)

for x in counter:
    print(x)
```

##### Decorators

functions that allows to modify function or behaviour of a function/class method without modifying their actual code

```python
# syntax and basic decorator
def decorator(func):

    def wrapper():
        print("Starting")
        func()
        print("Finished")

    return wrapper
```

```python
@decorator
def greet():
    print("Hello")
```

```python
def greet():
    print("Hello")

greet = decorator(greet)

# this and earlier column are same 
```

```python
greet()
```

```python
#closure functions - function inside a function
def main_welcome():
    msg="Welcome"
    def sub_welcome_method():
        print("WElcome to advanced python")
        print(msg) ##  possible to use main function varibale in sub function
        print("please learn concepts properly")

    return sub_welcome_method()
```

```python
main_welcome()
```

```python
#Decorator with arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3) # take 3 as parameter and say_hello as func in decorator 
def say_hello():
    print("Hello")


say_hello()
```

```python
#this preserves things like __name__, __doc__ 
from functools import wraps
def decorator(func):

    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)

    return wrapper
```
