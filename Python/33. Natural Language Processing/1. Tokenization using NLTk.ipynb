```python
corpus = """Hello! Welcome to Agam's portfolio. 
Please have a look at my resume and projects and contact regarding any project."""
```

```python
print(corpus)
```

```python
# Converting paragraphs into sentences
from nltk.tokenize import sent_tokenize
```

```python
# Additional library download for tokenization
import nltk
nltk.download('punkt_tab')
```

```python
documents= sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization - 
# paragraph into words
# Sentence into words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
# word tokenise on sentences
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
# tokenize punctuation also
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
# here full stop is not considered as seprate word except last full stop
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
