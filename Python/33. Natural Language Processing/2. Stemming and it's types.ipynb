#### Stemming
Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or
to the roots of words known as a lemma.

```python
# Classification Problem
# e.g. Comments of product is positive or negative review
# Reviews-> eating,eat,eaten [going,gone,goes]->go

words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

#### Porter stemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"----->"+stemming.stem(word))
```

```python
# disad: words may lost it's meaning -> fixed with help of lammatization
stemming.stem('congratulations')
```

```python
stemming.stem('sitting')
```

#### RegexpStemmer class

NLTK has RegexpStemmer class with the help of which we can easily implement Regular
Expression Stemmer algorithms. 

It basically takes a single regular expression and removes any prefix or suffix that matches the expression.

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ineating')
```

### Snowball Stemmer

stemming algorithm which is also known as the Porter2 stemming algorithm as it is a better version of the Porter Stemmer since some issues of it were fixed in this stemmer.

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"-->"+snowballstemmer.stem(word))
```

```python
stemming.stem("fairly"), stemming.stem("sportingly")
```

```python
# better performance
snowballstemmer.stem("fairly"), snowballstemmer.stem("sportingly")
```
