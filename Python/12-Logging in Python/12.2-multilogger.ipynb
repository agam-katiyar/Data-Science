##### LOgging with multiple loggers

can create multiple multiple loggers for different parts of your application.

```python
import logging

## create logger for module 1
logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)


## Logger for module 2
logger2 = logging.getLogger("modele2")
logger2.setLevel(logging.WARNING)

logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt = '%Y-%m-%d %H:%M:%S'
)
```

```python
## log message with different loggers and different modules

logger1.debug("THis is debug message for module1")
logger2.warning("THis is a warning message for module2")
logger2.error("THis is an error message")
```
