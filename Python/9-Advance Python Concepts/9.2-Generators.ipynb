Generators

Subclass of iterator

Simpler way to create iterators.

Use yield keyword to produce a series of values lazily(they genrerate values on fly and don't store them in memory)

```python
def square(n):
    for i in range(3):
        yield i**2 ## yield create local variable and save them also
```

```python
square(3)
```

```python
for i in square(3): # iteratively calls function
    print(i)
```

Generators are useful for reading large files because they allow you to process one line at a time without loading the entire file into memory

```python
# Practical eg. : Reading large files
# whole function is iterator
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path='large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
