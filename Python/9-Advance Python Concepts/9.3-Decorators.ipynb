## Decorators

Power and flexible feature in python

allows to modify behaviour of a function or class method

commonly used to add functionality to functions or methods without modifying their actual code

## Function copy

```python
def welcome():
    return "Welcome to home"

welcome()
```

```python
wel = welcome ## copied  welcome function
print(wel()) # calling copied function
del welcome
print(wel()) ## copied , deleting welcome don't affect wel
```

## Closures

Function  inside a function

```python
## Closure functions
def main_welcome():
    msg="Welcome"
    def sub_welcome_method():
        print("WElcome to advanced python")
        print(msg) ##  possible to use main function varibale in sub function
        print("please learn concepts properly")

    return sub_welcome_method()
```

```python
main_welcome()
```

```python
def main_welcome(func):
    def sub_welcome_method():
        print("WElcome to advanced python")
        func("Welcome everyone") 
        print("please learn concepts properly")

    return sub_welcome_method()
```

```python
main_welcome(print)  #Inbuilt print function
```

```python
def main_welcome(func,lst):
    def sub_welcome_method():
        print("WElcome to advanced python")
        print(func(lst))
        print("please learn concepts properly")

    return sub_welcome_method()
```

```python
main_welcome(len,[1,2,3,4,5,6])
```

# Decorators

```python
# decorator
def main_welcome(func):
    def sub_welcome_method():
        print("WElcome to advanced python")
        func() 
        print("please learn concepts properly")

    return sub_welcome_method()
```

```python
def course_Introduction():
    print("This is advanced python course")

course_Introduction()
```

```python
main_welcome(course_Introduction)
```

```python
@main_welcome  
# curse_introdutcion is going inside main_introduction as parameter
def course_Introduction():
    print("This is advanced python course")
#course_introduction as decoration function
```

```python
## Decroator
def my_decorator(func):
    def wrapper():
        print("something happeing before function is called")
        func()
        print("Something happening after function is called")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello")
```

```python
say_hello()
```

```python
#Decorator wirth arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3) # take 3 as parameter and say_hello as func in decorator 
def say_hello():
    print("Hello")
```

```python
say_hello()
```
