Iterators

Allows for efficient looping and memory management. 

provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list =[1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
print(my_list)
```

```python
## Iterator
iterator = iter(my_list)
print(type(iterator))
```

```python
iterator  ## lazy loading technique is used here 
```

```python
## Iterate through all the elements
next(iterator) ## Only 1 value by lazy loading
```

```python
iterator = iter(my_list)
```

```python
try:
    print(next(iterator))
except StopIteration: ## After 6(Values in list) iterations
    print("There are no elements in the iterator")
```

```python
# String iterator
my_string = "AGAM"
string_iterator=iter(my_string)

print(next(string_iterator))
print(next(string_iterator))
```
