Functions

Block of code that performs a specific code.

```python
##Syntax
def function_name(parameteres):
    """Docstring:""" # write comments about function
    # Function body
    return expression
```

```python
## Why functions? TO improve code reusability

def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("Number is even")
    else:
        print("Number is odd")

## calling the function
even_or_odd(10)
```

```python
## Function with multiple parameter
def add(a,b):
    c=a+b
    return c

add(4,5)
```

```python
## Default parameters

def greet(name="Guest"):
    print(f"Hello {name}")

greet()
greet("agam")
```

```python
## Variable length arguments
## Positional and keywords arguments


# Positional arguments
# Multiple arguments 
def print_numbers(*args): # we can write anything instead of "args"
    for numbers in args:
        print(numbers)

print_numbers(1,3,5,6,4,7,8,9)
```

```python
# Keywords arguments
# All the arguments will be in the form of key value pairs
def print_detail(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_detail(name="Agam",age=32,country="india")
```

```python
def print_detail(*args,**kwargs):
    for val in args:
        print(f" Positional arguments :{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_detail(1,2,3,4,name="Agam",age=32,country="india")
```

```python
## return statement
def multiply(a,b):
    return a*b

multiply(4,6)
```

```python
## Return multiple parameters
def multiply(a,b):
    return a*b,a

multiply(4,5)
```
