map() Function 

It applies a given function to all items in an input list(or any other literals) and returns a map object(an iterator). This is particularly useful for transforming data in a list comprehensively

takes two input parameters: function name and iterable(collection)

```python
def square(x):
    return x*x
square(4)
```

```python
numbers=[1,2,3,4,5,6,7,8]

list(map(square,numbers))
```

```python
## Lambda function with map
numbers=[1,2,3,4,5,6,7,8]
list(map(lambda x:x**2,numbers))
```

```python
## Map multiple iterables

numbers1=[1,2,3]
numbers2=[4,5,6]

added_numbers=list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```

```python
## map() to convert list of string to integers
## Use map to convert strings to integers
str_numbers=['1','2','3','4','5']
int_numbers=list(map(int, str_numbers))

print(int_numbers)
```

```python
## Inbuilt function with help of map()
words=['apple','banana','cherry']
upper_words=list(map(str.upper,words))
print(upper_words)
```

```python
def get_name(person):
    return person['name']

people=[
    {'name':'krish','age':23},
    {'name':'jac','age':45}   
]

list(map(get_name,people))
```
