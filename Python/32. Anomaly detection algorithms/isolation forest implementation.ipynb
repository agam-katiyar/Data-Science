#### Isolation Forest on healthcare dataset

```python
import pandas as pd
df = pd.read_csv('healthcare.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.iloc[:,0],df.iloc[:,1])
```

Implementation

```python
from sklearn.ensemble import IsolationForest
```

```python
clf = IsolationForest(contamination=0.2) ## contaminationn= allowed/threshold value of anomaly
clf.fit(df)
predictions = clf.predict(df)
```

```python
predictions
```

```python
import numpy as np
index = np.where(predictions < 0) # index of outliers
index
```

```python
x= df.values
```

```python
index = np.where(predictions< 0)
plt.scatter(df.iloc[:,0], df.iloc[:,1])
plt.scatter(x[index,0], x[index,1], edgecolors="r")
```
