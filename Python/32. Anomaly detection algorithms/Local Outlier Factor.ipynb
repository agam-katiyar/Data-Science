```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.datasets import make_circles
```

```python
X, y= make_circles(n_samples=900, factor=0.2, noise=0.1)
```

```python
plt.scatter(X[:,0],X[:,1], c= y)
```

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
lof = LocalOutlierFactor(n_neighbors=3, algorithm='kd_tree',metric='minkowski')
```

```python
lof
```

```python
y_pred = lof.fit_predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], c= y_pred)
```
