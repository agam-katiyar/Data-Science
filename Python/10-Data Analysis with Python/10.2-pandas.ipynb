# Pandas

Powerful data manipulation library.
Used for data analysis and cleaning.
Provides 2 primary data structures: Series and dataframe

Series: One dimensional array like object that can hold any data type. 
similar to a column in a table

dataframe: 2D, size-mutable and potentially heterogeneous tabular data structure with labelled axes(rows & columns)

```python
import pandas as pd
```

# Series

```python
## Series
import pandas as pd
data=[1,2,3,4,5]
series = pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
# create Series from dictionary
data= {'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

## Dataframe

```python
# Dataframe 
# Create a dataframe from dictionary of list

data={
    'Name':['Krish','Agam','John'],
    'Age': [20,25,45],
    'City': ['Bangalore','New York','Kanpur']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
# Create data frame from  a list of dictionaries
data=[
    {'Name':'Krish','Age':32,'City':'Banglore'},
    {'Name':'Agam','Age':22,'City':'Kanpur'},
    {'Name':'Krish','Age':32,'City':'Banglore'},
    {'Name':'Krish','Age':32,'City':'Banglore'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
