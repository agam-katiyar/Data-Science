Data Manipulation and Analysis with Pandas

Pandas provide a wide range of functions for data manipulation and analsyis

```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
## Fetch first 5 rows
df.head(5)
```

```python
# Handling missing values
df.isnull().any()
```

```python
# Calculating missing values columnwise
df.isnull().sum()
```

```python
# Replace missing values with 0
df_filled=df.fillna(0)
```

```python
## Filling missing values with the mean of the values (NAN-> Not a Number)
df['Stock_fillNA']=df['Stock'].fillna(df['Stock'].mean())
df
```

```python
df.dtypes #Object->means String
```

```python
# Renaming columns
df=df.rename(columns={'Price':'Sales Price'})
df.head()
```

```python
#CHanging data types
df['Sales Price New']=df['Sales Price'].astype(float)
df.head()
```

```python
# Modifying data
df['New Sales Price']=df['Sales Price New'].apply(lambda x:x*2)
df.head()
```

```python
# Data aggregating and grouping
df.head()
```

```python
# Grouping based on product
grouped_mean=df.groupby('Size')['Stock'].mean() # grouppping based on size and printing mean of stocks
print(grouped_mean)
```

```python
group_sum=df.groupby(['Size','Color'])['Stock'].sum()
print(group_sum)
```

```python
group_mean=df.groupby(['Size','Color'])['Stock'].mean()
print(group_mean)
```

```python
# Aggregate mutiple functions
grouped_agg=df.groupby('Color')['Stock'].agg(['mean','sum','count'])
grouped_agg
```

```python
## Merging and joining data frames
# Create sample dataframes
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
# Merge dataframe on the'Key' columns
pd.merge(df1,df2,on="Key",how="inner") #'on ' defines which column , 'how' defines which join
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
