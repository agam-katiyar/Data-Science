```python
import seaborn as sns
```

```python
## Basic plotting

tips = sns.load_dataset('tips')
print(tips)
```

```python
## creating a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip', data = tips)
plt.title("Scatter Plot of total bill vs Tip")
plt.show()
```

```python
## Line plot
sns.lineplot(x='size', y='total_bill', data = tips)
plt.title("Line plot of total bill by size")
plt.show()
```

```python
## Categorical plot

## bar plot
sns.barplot(x='day', y='total_bill', data = tips)
plt.title('Bar plot of total bill by day')
plt.show()
```

```python
## Box plot

sns.boxplot(x='day', y='total_bill',data = tips)
```

```python
#Violin plot
sns.violinplot(x='day',y='total_bill', data = tips)
```

```python
## Histogram
## KDE ->Kernel density estimation( shows curve(smoothens the histogram))
sns.histplot(tips['total_bill'], bins = 10, kde = True)
```

```python
# KDE plot

sns.kdeplot(tips['total_bill'], fill= True)
```

```python
# Pair plot
sns.pairplot(tips) # only numerical variables are considered
```

```python
## Heat map
corr=tips[['total_bill', 'tip','size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
# Customization of plot
import pandas as pd
data_df = pd.read_csv('data.csv')
data_df.head(5)
```

```python
## Plot total stock by size
plt.figure(figsize=(10,6))
sns.barplot(x='Size', y='Stock', data = data_df, estimator = sum)
plt.title("TOtal stock by size")
plt.xlabel("Category")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Availability', y='Stock', data = data_df, estimator = sum)
plt.title("TOtal stock by availability")
plt.xlabel("Category")
plt.show()
```
