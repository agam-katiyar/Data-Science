####  Memory Management in Python

 involves a combination of automatic garbage collection, reference counting,
and various internal optimizations to efficiently manage memory allocation and deallocation. Understanding
these mechanisms can help developers write more efficient and robust applications.

##### Reference counting

primary method pythonn uses to manage memory

each object in python maintains a count of references pointing to it. When reference count drops to zero, the memory occupied by the object is deallocated.

```python
import sys

a=[]

## Output: 2 (1 reference from 'a' and 1 from getrefcount())
print(sys.getrefcount(a))
```

```python
b=a
print(sys.getrefcount(b))
```

```python
print(sys.getrefcount(a)) # b reference is also added
```

```python
del b
print(sys.getrefcount(b))
```

```python
print(sys.getrefcount(a)) # b is deleted 
```

##### Garbage collection
python includes a cyclice garbage collector to handle refernce cycles. Reference cycles occur when objects reference each other, preventing their reference counts from reaching zero.

```python
import gc

## enable garbage collection
gc.enable()
```

```python
gc.disable()
```

```python
gc.collect()
```

```python
## garbage collection stats

print(gc.get_stats())
```

```python
## get unreachable  stats
print(gc.garbage)
```

// Memory Management Best Practices
1. Use Local Variables: Local variables have a shorter lifespan and are freed sooner than global
variables.
2. Avoid Circular References: Circular references can lead to memory leaks if not properly managed.
3. Use Generators: Generators produce items one at a time and only keep one item in memory at a
time, making them memory efficient.
4. Explicitly Delete Objects: Use the del statement to delete variables and objects explicitly.
5. Profile Memory Usage: Use memory profiling tools like tracemalloc and memory_profiler to identify
memory leaks and optimize memory usage.

```python
## Handling circular reference
import gc

class MyObject:
    def __init__(self,name):
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"{self.name} deleted")


## Create circular reference
obj1 = MyObject("obj1")
obj2 = MyObject("obj2")

obj1.ref = obj2
obj2.ref = obj1

#since we are giving circular reference  so it will not automatically collect, we have to manually collect garbage
del obj1
del obj2

## Manually trigger garbage collection
gc.collect()
```

```python
## Generators for memory efficiency
# allows to produce items one at a time, using memory efficiently

def generate_numbers(n):
    for i in range(n):
        yield i

## using generator
for num in generate_numbers(100000):
    print(num)
    if(num> 10):
        break
```

```python
## Profiling memory usage with tracemalloc

import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')

    print("[ Top 10 ]")
    for stat in top_stats[::]:
        print(stat)
```

```python
main()
```
