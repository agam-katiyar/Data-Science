Exceptions

Excceptions are events that disrupt the normal flow of a program. THey occur then an error is encountered during program execution.
eg.
1. ZeroDivisionError : Dividing by zero
2. FileNotFoundError : File not found
3. ValueError : Invalid value
4. TypeError : Invalid type

```python
# Exception try,except block
a=10
try:
    a=b
except:
    print("The variable has not been defined")
```

```python
## NameError exception
a=10
try:
    a=b
except NameError as ex:
    print(ex)
```

```python
try:
    result=1/2
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter denominator grater than 0")
```

```python
try:
    result=1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter denominator grater than 0")

## Parent class of exception(Should be in the last of code)
except Exception as ex1:
    print(ex1)
    print("Main exception got caught here")
```

```python
try:
    num = int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than 0")
except Exception as ex:
    print(ex)
```

```python
## try, except, else block
# Else block is executed if no exception is occured
try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than 0")
except Exception as ex:
    print(ex)
else:
    print(f"hey the result is {result}")
```

```python
## try,except,else, and finally 

##Finally block will be always executed in every condition- if error occur or not 
try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than 0")
except Exception as ex:
    print(ex)
else:
    print(f"hey the result is {result}")
finally:
    print("Execution complete")
```
