```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
# Importing IRIS dataset
iris = datasets.load_iris()
```

```python
iris
```

```python
# converting to dataframe
iris_data = pd.DataFrame(iris.data)
```

```python
iris_data.columns = iris.feature_names ## assigning features names to columns
```

```python
iris_data
```

#### Standard scaling- feature scaling

```python
# Standardization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(iris_data)
```

```python
X_scaled
```

```python
## Applying PCA to reduce the dimensions
from sklearn.decomposition import PCA
```

```python
# COnverting 4 features into 2 features
pca = PCA(n_components=2)
```

```python
pca_scaled = pca.fit_transform(X_scaled)
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1], c= iris.target) # color = target valuse in dataset
```

#### Agglomerative clustering

```python
## Constructing dendogram
import scipy.cluster.hierarchy as sc

## Plotting dendogram
plt.figure(figsize=(20,7))
plt.title("Dendograms")

## Creating dendogram
sc.dendrogram(sc.linkage(pca_scaled, method='ward'))
plt.title("dendogram")
plt.xlabel('Sample index')
plt.ylabel('Euclidean distance') # by default euclidean distance is used in agglomerative
```

```python
## taking k=2
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=2, metric= 'euclidean', linkage='ward')
cluster.fit(iris_data)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1],c=cluster.labels_)
```
