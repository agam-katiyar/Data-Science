Importing Modules in Python

Modules and packages help organize and reuse codes.

```python
import math
math.sqrt(9)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
from math import *
print(sqrt(36))
print(pi)
```

```python
## Custom module 
from package.maths import addition
addition(3,4)
```
