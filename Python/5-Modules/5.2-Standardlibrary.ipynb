```python
## Data serialiazation
import json
data={'name':'agam','age':21}

json_str=json.dumps(data) ## Convert to json
print(json_str) 
print(type(json_str))

parsed_data = json.loads(json_str) ##convert to dictionary
print(parsed_data)
print(type(parsed_data))
```

```python
## csv
import csv

with open('example.csv', mode='w', newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['agam',21])

with open('example.csv', mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## data time
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)
print(yesterday)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## Regular expression

import re

pattern=r'\d+'  ## d=digits
text='There are 123 apples 346'
match=re.search(pattern,text)
print(match.group())
```
