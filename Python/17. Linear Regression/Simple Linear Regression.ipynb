Regression

```python
## INdependent and dependent features
X=df[['Weight']] # Independent feature should be dataframe or 2 dimensional array
Y=df['Height'] # This dependent feature can be in series or  1d array
type(X)
np.array(X).shape
```

```python
# One d array - will give error
X_series=df['Weight']
np.array(X_series).shape
```

```python
np.array(Y).shape
```

```python
## Train test split
from sklearn.model_selection import train_test_split
 
```

```python
X_train,X_test, Y_train,Y_test = train_test_split(X,Y,test_size=0.25,random_state=42)
```

```python
X_train.shape
```

```python
## Standardization- if we are using values of different unit
#We take each independent feature and apply z score- converting all values with mean = 0, standard deviation = 1
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train=scaler.fit_transform(X_train) #Use training dataset mean and standard deviation here
```

```python
X_test= scaler.transform(X_test) #Use training dataset mean and standard deviation here to prevent data leakage
```

```python
X_test
```

```python
## Apply simple linear regression
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression(n_jobs=-1)
```

```python
## Error while using 1 d array or series
regression.fit(X_series,Y)
```

```python
regression.fit(X_train,Y_train)
```

```python
print("Coeeficient or slope",regression.coef_) # 1 unint movement in weight unit makes 17.29 movement in height
print("Intercept :",regression.intercept_) # when weight is 0
```

```python
# Plot training data plot best fit line
plt.scatter(X_train,Y_train)
plt.plot(X_train,regression.predict(X_train))
```

#### Prediction of test data
1. predicted height output = intercept + coef_(Weights)
2. y_pred_test = 15647.2 + 17.29(X_test)

```python
# Prediction for test data
y_pred = regression.predict(X_test)
```

```python
# Performance metrix
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
mse = mean_squared_error(Y_test,y_pred)
mae = mean_absolute_error(Y_test, y_pred)
rmse = np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

#### R square
Formula 
R^2 = 1- SSR/SST
R^2 = coeeficient of deteremination ,
SSR = sum pf squares of residuals ,
SST = total sum of squares 

```python
from sklearn.metrics import r2_score
```

```python
score = r2_score(Y_test,y_pred)
print(score)
```

#### Adjusted r square
Formula 
a-[(1-R2)*(n-1)/(n-k-1)]

where : 
R2 : the r2 of the model ,
n: the number of observations,
k : the number of predictor variables

```python
# DIsplay adjusted r square
1-(1-score)*(len(Y_test)-1)/(len(Y_test)-X_test.shape[1]-1)
```
