```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
df_index = pd.read_csv("economic_index.csv")
```

```python
df_index.head()
```

```python
# Drop unneccessary columns
df_index.drop(columns = ["Unnamed: 0","year","month"],axis =1,inplace=True)
```

```python
df_index.head()
```

```python
## CHeck null values
df_index.isnull().sum()
```

```python
# Visualization
import seaborn as sns
sns.pairplot(df_index)
```

```python
df_index.corr()
```

```python
## Visualixe datapoints more closely
plt.scatter(df_index['interest_rate'],df_index['unemployment_rate'],color='b')
plt.xlabel("Interest rate")
plt.ylabel("Unemployment rate")
```

```python
## INdependent and dependent features
X= df_index.iloc[:,:-1] # iloc= index locate, all rows , just remove last row
Y = df_index.iloc[:,-1] 
```

```python
X.head()
```

```python
Y
```

```python
# Train test split
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.25,random_state=42)
```

```python
import seaborn as sns
```

```python
## REG plot
sns.regplot(x=df_index['interest_rate'],
            y=df_index['index_price']
    )
```

```python
sns.regplot(x=df_index['interest_rate'],
            y=df_index['unemployment_rate']
    )
```

```python
sns.regplot(x=df_index['index_price'],
            y=df_index['unemployment_rate']
    )
```

```python
# Standardizing the data
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
```

```python
regression.fit(X_train,Y_train)
```

```python
# Cross validation of the model
from sklearn.model_selection import cross_val_score
validation_score = cross_val_score(regression,X_train,Y_train,scoring='neg_mean_squared_error',cv=3) # different scoring paramteres available(refer sklearn website),cv = no of cross validation 
```

```python
validation_score
```

```python
np.mean(validation_score)
```

```python
## Prediction
Y_pred = regression.predict(X_test)
```

```python
Y_pred
```

```python
## Performance metrices
from sklearn.metrics import mean_absolute_error, mean_squared_error
mse = mean_squared_error(Y_test,Y_pred)
mae = mean_absolute_error(Y_test,Y_pred)
rmse = np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

```python
# r square and adjusted r square
from sklearn.metrics import r2_score
score = r2_score(Y_test,Y_pred)
print(score)

## adjusted r square
print(1 - (1-score)*(len(Y_test)-1)/(len(Y_test)-X_test.shape[1]-1))
```
