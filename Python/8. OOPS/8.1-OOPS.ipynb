```python
## Instance variable and methods
class Dog:
    ## Constructor
    def __init__(self,name, age): ## Self keyword is responsible for accesing the instance variable 
        self.name=name #Instance variable
        self.age=age

## Creating object
dog1=Dog("Buddy",3)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2= Dog("Lucy",2)
print(dog2.name)
print(dog2.age)
```

```python
## Instance methods
## DEfine a class with instance method
class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def bark(self): #Instance method
        print(f"{self.name} says woof")

dog1 = Dog("Buddy",2)
dog1.bark()
dog2 = Dog("Lucy",4)
dog2.bark()
```

```python
## MOdelling a bank account

## Define a class for bank account
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance
    
    def depost(self,amount):
        self.balance+=amount
        print(f"{amount}  is deposited. New balance is {self.balance} ")

    def withdraw(self,amount):
        if(amount>self.balance):
            print("Insufficient funds")
        else:
            self.balance-=amount
            print(f"Rupees {amount} is withdrawn. New Balance is {self.balance}")

    def get_balance(self):
        return self.balance
    

## CReate an account
account=BankAccount("Agam",3000)
print(account.balance)

```

```python
## call instance method
account.depost(1000)
```

```python
account.withdraw(500)
```

```python
print(account.get_balance())
```
