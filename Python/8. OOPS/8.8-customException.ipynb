Custom exception(Raise and throw an exception)

```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year =int(input("enter date of birth"))
age=2026-year
try:
    if age<=30 and age>=20:
        print("Age is valid and you can apply for exam")
    else:
        raise dobException
except dobException:
    print("Your age does not fall in between 20 and 30")
```
