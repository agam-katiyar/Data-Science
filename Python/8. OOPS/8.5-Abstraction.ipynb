Abstraction

concept of hiding the complex implementation details and showing only necesssary features of an obejct.

Reduces complexity and effort

```python
from abc import ABC, abstractmethod

## abstract claas
class Vehicle(ABC):
    def drive(self):
        print("Vehicle is used for driving")


    ## abstract method is implemented at other derived class 
    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):
    def start_engine(self):
        print("car engine started")

def operate_vehicle(vehicle):
    vehicle.start_engine()

car=Car()
operate_vehicle(car)
```
