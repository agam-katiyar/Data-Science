Operators

```python
##Arithmetic operations

a=45
b=23

addition=a+b
print(addition)

subtraction=a-b
print(subtraction)

multiplication=a*b
print(multiplication)

division=a/b
print(division)

floordivision=a//b
print(floordivision)

modulus=a%b
print(modulus)

exponential =a**b
print(exponential)
```
